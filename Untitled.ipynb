```python
import os, httpx
from dotenv import load_dotenv

load_dotenv()

GROQ_URL = "https://api.groq.com/openai/v1/chat/completions"
MODEL = os.getenv("LLM_MODEL", "openai/gpt-oss-120b")

def ask_llm(question: str, system: str = "أنت مساعد مفيد يجيب بدقة وإيجاز.") -> str:
    r = httpx.post(
        GROQ_URL,
        headers={"Authorization": f"Bearer {os.getenv('GROQ_API_KEY')}"},
        json={
            "model": MODEL,
            "messages": [
                {"role": "system", "content": system},
                {"role": "user", "content": question},
            ],
            "temperature": 0.2,
        },
        timeout=60,
    )
    if r.status_code != 200:
        raise RuntimeError(f"LLM error {r.status_code}: {r.text}")
    return r.json()["choices"][0]["message"]["content"]


print(ask_llm("ما الفرق بين SQL و NoSQL؟ في ثلاث نقاط."))
```

```python
def split_text(text: str, chunk_size: int = 500, overlap: int = 100) -> list[str]:
    chunks = []
    start = 0
    while start < len(text):
        end = start + chunk_size
        chunks.append(text[start:end])
        start = end - overlap
    return chunks


chunks = split_text(text, chunk_size=80, overlap=20)

print("عدد القطع:", len(chunks))
for i, c in enumerate(chunks):
    print(f"\n--- القطعة {i} ---")
    print(repr(c))
```

```python
from sentence_transformers import util

query = "كم يوم أقدر أغيب لو مرضت؟"

candidates = [
    "Sick leave: 15 days per year.",
    "Annual leave: 30 days per year.",
    "The company cafeteria opens at 7 AM.",
]

q_vec = model.encode(query)
c_vecs = model.encode(candidates)

scores = util.cos_sim(q_vec, c_vecs)[0]

for text, score in zip(candidates, scores):
    print(f"{score:.4f}  |  {text}")
```

```python
import os
import httpx
import chromadb
from chromadb.utils import embedding_functions
from dotenv import load_dotenv

load_dotenv()

GROQ_URL = "https://api.groq.com/openai/v1/chat/completions"
MODEL = os.getenv("LLM_MODEL", "openai/gpt-oss-120b")

client_db = chromadb.PersistentClient(path="./vectorstore")
embed_fn = embedding_functions.SentenceTransformerEmbeddingFunction(
    model_name="sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2"
)
collection = client_db.get_collection("company_docs", embedding_function=embed_fn)


SYSTEM_PROMPT = """أنت مساعد مؤسسي. أجب على السؤال بالاعتماد على السياق المرفق فقط.

قواعد إلزامية:
- لا تستخدم أي معرفة خارج السياق المرفق.
- إذا لم يحتوِ السياق على الإجابة، قل بوضوح: "لا تحتوي المستندات المتاحة على إجابة لهذا السؤال."
- لا تخمّن ولا تستنتج معلومات غير مذكورة صراحة.
- أشر إلى رقم المصدر الذي اعتمدت عليه، مثل [مصدر 1].
- أجب بإيجاز."""


def retrieve(question: str, k: int = 3) -> list[dict]:
    res = collection.query(query_texts=[question], n_results=k)
    return [
        {"text": doc, "meta": meta, "distance": dist}
        for doc, meta, dist in zip(
            res["documents"][0], res["metadatas"][0], res["distances"][0]
        )
    ]


def build_context(chunks: list[dict]) -> str:
    parts = []
    for i, c in enumerate(chunks, start=1):
        m = c["meta"]
        parts.append(f"[مصدر {i}] (ملف: {m['source']}، صفحة: {m['page']})\n{c['text']}")
    return "\n\n".join(parts)


def ask_rag(question: str, k: int = 3) -> dict:
    chunks = retrieve(question, k)
    context = build_context(chunks)

    user_msg = f"السياق:\n{context}\n\nالسؤال: {question}"

    r = httpx.post(
        GROQ_URL,
        headers={"Authorization": f"Bearer {os.getenv('GROQ_API_KEY')}"},
        json={
            "model": MODEL,
            "messages": [
                {"role": "system", "content": SYSTEM_PROMPT},
                {"role": "user", "content": user_msg},
            ],
            "temperature": 0.0,
        },
        timeout=60,
    )
    if r.status_code != 200:
        raise RuntimeError(f"LLM error {r.status_code}: {r.text}")

    return {
        "answer": r.json()["choices"][0]["message"]["content"],
        "sources": chunks,
    }
```

```python
out = ask_rag("كم يوم إجازة مرضية لدي؟")

print(out["answer"])
print("\n" + "=" * 50)
print("المصادر المستخدمة:\n")
for i, s in enumerate(out["sources"], start=1):
    print(f"[مصدر {i}] {s['meta']['source']} — صفحة {s['meta']['page']}")
    print(f"   المسافة: {s['distance']:.3f}")
    print(f"   {s['text']}\n")
```

```python
def retrieve(question: str, k: int = 3, max_distance: float = 0.6) -> list[dict]:
    res = collection.query(query_texts=[question], n_results=k)
    chunks = [
        {"text": doc, "meta": meta, "distance": dist}
        for doc, meta, dist in zip(
            res["documents"][0], res["metadatas"][0], res["distances"][0]
        )
    ]
    return [c for c in chunks if c["distance"] <= max_distance]
    
```

```python
def ask_rag(question: str, k: int = 3, max_distance: float = 0.6) -> dict:
    chunks = retrieve(question, k, max_distance)

    if not chunks:
        return {
            "answer": "لا تحتوي المستندات المتاحة على معلومات ذات صلة بهذا السؤال.",
            "sources": [],
        }

    context = build_context(chunks)
    user_msg = f"السياق:\n{context}\n\nالسؤال: {question}"

    r = httpx.post(
        GROQ_URL,
        headers={"Authorization": f"Bearer {os.getenv('GROQ_API_KEY')}"},
        json={
            "model": MODEL,
            "messages": [
                {"role": "system", "content": SYSTEM_PROMPT},
                {"role": "user", "content": user_msg},
            ],
            "temperature": 0.0,
        },
        timeout=60,
    )
    if r.status_code != 200:
        raise RuntimeError(f"LLM error {r.status_code}: {r.text}")

    return {
        "answer": r.json()["choices"][0]["message"]["content"],
        "sources": chunks,
    }
```

```python
for q in [
    "كم يوم إجازة مرضية؟",
    "متى يفتح المقصف؟",
    "ما سياسة العمل عن بعد؟",
]:
    out = ask_rag(q)
    print(f"❓ {q}")
    print(f"💬 {out['answer']}")
    print(f"📎 عدد المصادر: {len(out['sources'])}\n{'-'*50}")
```
